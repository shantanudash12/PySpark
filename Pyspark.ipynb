```python
df = spark.read.format("csv").option("header","true").load("/Volumes/pyspark/default/pysparkfiles/original.csv")
```

```python
display(df)
```

```python
from pyspark.sql.functions import *
```

```python
df2 = df.withColumn("clean_city", when(df.City.isNull(), "Unknown").otherwise(df.City))
```

```python
display(df2)
```

```python
df2 = df2.filter(df2.JobTitle.isNotNull())
display(df2)
```

```python
df2 = df2.withColumn("clean_salary", round(df2.Salary.substr(2,100).cast("double"), 2))
display(df2)
```

```python
mean = df2.groupBy().avg("clean_salary").take(1)[0][0]
```

```python
print(mean)
```

```python
from pyspark.sql.functions import lit
df2 = df2.withColumn("new_salary", when(df2.clean_salary.isNull(),lit(mean)).otherwise(df2.clean_salary))
display(df2)
```
